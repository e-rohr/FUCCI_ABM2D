```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import joypy
```

```python
np.random.seed(0)
X = np.random.uniform(size=(500,2))

c = ( 1*(X[:,1] <= 1/3) 
     + 2*(np.logical_and(X[:,1]>1/3, X[:,1] <= 2/3))
     + 3*(X[:,1] > 2/3) )

plt.figure(layout="constrained")

scatter = plt.scatter(*X.T,c=c,edgecolors="k",s=50)

fontsize=22

### To create legend
# legend1 = plt.legend(*scatter.legend_elements(),
#                     title="Clusters",fontsize=fontsize,ncols=3)
# plt.setp(legend1.get_title(), fontsize=fontsize)
# plt.axis([1,2,1,2])

plt.xticks([0,1/3,2/3,1],["0","1/3","2/3","1"],fontsize=fontsize)
plt.yticks([0,1/3,2/3,1],["0","1/3","2/3","1"],fontsize=fontsize)

plt.title("Uniform data for clustering 1",fontsize=fontsize)
plt.xlabel("$x_1$",fontsize=fontsize)
plt.ylabel("$x_2$",fontsize=fontsize)

plt.savefig("../figures/ridgeline_clusters.pdf",format="pdf");
```

```python
cluster_distns = []
for i in range(1,4):
    cluster_distn = {}
    cluster_distn = {r"$x_1$": X[c==i,0], r"$x_2$": X[c==i,1]}
    cluster_distns.append(cluster_distn)

# Using 1/φ as the overlap parameter for ridgeline plots
o = (1+np.sqrt(5))/2
distns_plot = []

distn_plots = []
for i in range(3):
    distn_plot = {}
    distn_plot[r'$x_1$'] = cluster_distns[i][r'$x_1$']
    distn_plot[r'$x_2$'] = cluster_distns[i][r'$x_2$']
    distn_plots.append(distn_plot)
```

```python
for i in range(3):
    fig, ax = joypy.joyplot(distn_plots[i],
                            colormap = mpl.colormaps['cool'].resampled(2),
                            figsize = (6,2),
                            linecolor='k',#'#f0f0f0',
                            grid = 'y',
                            ylabelsize = 20,
                            xlabelsize = 20,
                            ylim ='own',
                            overlap = 0.5,
                            fade=True,
                            linewidth = 3)
    
    ax[-1].set_xlabel("Parameter values",fontsize=20)
    #ax[-1].set_title(f"Cluster {i+1} ridgeline plot",fontsize=20)
    
    plt.savefig(f"../figures/ridgeline_clusters_{i}.pdf",format="pdf");
```

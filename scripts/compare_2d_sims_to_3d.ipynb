```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors
import src.abm2d as abm2d
import scipy.io
```

```python
### import 3d cross section
mat = scipy.io.loadmat('../data/3d_simulation/baseline_3d_sim.mat')
t = mat["t"].T
red = mat["redcounts"]
gre = mat["grecounts"]
yel = mat["yelcounts"]
dead = mat["deadcounts"]
```

## Plot 2d simulations while varying N(0)

```python
fig, axes = plt.subplots(2,2, figsize = (10,10),layout="constrained")
fontsize = 20

N = 1100
alpha = 0.01
for N in [750,850,950,1100,1250]:

    L = 1000
    args = {'N':N,
            'α':alpha,
            'L':L,
        'path': f"../data",
        'title': f"abm2d_alpha_{alpha}_N_{N}",
       }
    out = np.load(f"{args['path']}/{args['title']}.npz", allow_pickle=True)
    
    Nr = out['Nr']
    Ny = out['Ny']
    Ng = out['Ng']
    Nd = out['Nd']
    
    print(f"N = {N}")
    print(np.linalg.norm(Nr-red))
    
    axes[0,0].plot(np.arange(0,241)/24, Nr,linewidth = 3,  label = fr'$N$ = {N}')
    axes[0,1].plot(np.arange(0,241)/24, Ny,linewidth = 3,   label = fr'$N$ = {N}')
    axes[1,0].plot(np.arange(0,241)/24, Ng,linewidth = 3,   label = fr'$N$ = {N}')
    axes[1,1].plot(np.arange(0,241)/24, Nd,linewidth = 3,   label = fr'$N$ = {N}')

axes[0,0].plot(t,red,"ks--",markersize=10,label = "3D Klowss Model")
axes[1,0].plot(t,gre,"ks--",markersize=10)
axes[0,1].plot(t,yel,"ks--",markersize=10)
axes[1,1].plot(t,dead,"ks--",markersize=10)
    
ax = axes[0,0]    
ax.set_xlim(0,10)
ax.set_xticks([0,5,10],[0,5,10],fontsize=fontsize)
ax.set_ylim(0,4300)
ax.set_yticks([0,2e3,4e3],[0,2000,4000],fontsize=fontsize,rotation=90)
ax.grid(axis="y",linestyle="--")
ax.set_title("G1 cell counts",fontsize=fontsize)    
ax.set_ylabel("Cell counts",fontsize=fontsize)    
axes[0,0].legend(fontsize=fontsize*.75)

ax = axes[0,1]    
ax.set_xlim(0,10)
ax.set_xticks([0,5,10],[0,5,10],fontsize=fontsize)
ax.set_ylim(0,75)
ax.set_yticks([0,30,60],[0,30,60],fontsize=fontsize,rotation=90)
ax.grid(axis="y",linestyle="--")
ax.set_title("Early S cell counts",fontsize=fontsize)    

ax = axes[1,0]    
ax.set_xlim(0,10)
ax.set_xticks([0,5,10],[0,5,10],fontsize=fontsize)
ax.set_ylim(0,420)
ax.set_yticks([0,200,400],[0,200,400],fontsize=fontsize,rotation=90)
ax.grid(axis="y",linestyle="--")
ax.set_title("S/G2/M cell counts",fontsize=fontsize)    
ax.set_ylabel("Cell counts",fontsize=fontsize)    
ax.set_xlabel("Time (days)",fontsize=fontsize)    

ax = axes[1,1]    
ax.set_xlim(0,10)
ax.set_xticks([0,5,10],[0,5,10],fontsize=fontsize)
ax.set_ylim(0,1500)
ax.set_yticks([0,600,1200],[0,600,1200],fontsize=fontsize,rotation=90)
ax.grid(axis="y",linestyle="--")
ax.set_title("Dead cell counts",fontsize=fontsize)    
ax.set_xlabel("Time (days)",fontsize=fontsize) 

plt.savefig("../figures/model_2d_3d_comparison_varying_N.pdf",format="pdf")
```

## Plot 2d simulations while varying alpha

```python
import os, re

# Folder containing your files
folder = "../data/dataset1/"

# Pattern to extract float values and iteration
pattern = r"c_a=([0-9.eE+-]+)_eta1=([0-9.eE+-]+)_itr=(\d+)\.npz"

for fname in os.listdir(folder):
    if fname.endswith(".npz"):
        match = re.match(pattern, fname)
        if match:
            c_a = round(float(match.group(1)), 5)
            eta1 = round(float(match.group(2)), 5)
            itr = int(match.group(3))

            # Construct new filename with rounded values
            new_fname = f"c_a={c_a:.5f}_eta1={eta1:.5f}_itr={itr}.npz"

            # Perform rename
            src = os.path.join(folder, fname)
            dst = os.path.join(folder, new_fname)

            # Only rename if names are different
            if src != dst:
                os.rename(src, dst)
                print(f"Renamed: {fname} → {new_fname}")
```
